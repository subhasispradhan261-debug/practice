```

# ==========================================
# MODULE 01: INTRODUCTION TO AI & PYTHON
# ==========================================

print("===== INTRODUCTION TO AI & PYTHON =====")

# 1. AI, ML AND DEEP LEARNING

print("\n1. AI, ML AND DEEP LEARNING")

print("AI: Artificial Intelligence")
print("ML: Machine Learning")
print("DL: Deep Learning")

print("AI is the broader field.")
print("ML is a subset of AI.")
print("DL is a subset of ML.")


# 2. VARIABLES

print("\n2. VARIABLES")

name = "Subhasis"
age = 22
course = "Artificial Intelligence"
salary = 30000

print("Name:", name)
print("Age:", age)
print("Course:", course)
print("Salary:", salary)


# 3. DATA TYPES

print("\n3. DATA TYPES")

a = 100
b = 10.5
c = "Python"
d = True

print(type(a))
print(type(b))
print(type(c))
print(type(d))


# 4. OPERATORS

print("\n4. OPERATORS")

x = 20
y = 10

print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Modulus:", x % y)


# 5. IF ELSE CONDITION

print("\n5. IF ELSE CONDITION")

marks = 75

if marks >= 80:
    print("Grade A")

elif marks >= 60:
    print("Grade B")

else:
    print("Grade C")


# 6. FOR LOOP

print("\n6. FOR LOOP")

for i in range(1, 6):
    print(i)


# 7. WHILE LOOP

print("\n7. WHILE LOOP")

number = 1

while number <= 5:
    print(number)
    number += 1


# 8. LISTS

print("\n8. LISTS")

numbers = [10, 20, 30, 40, 50]

print("List:", numbers)
print("First Value:", numbers[0])
print("Last Value:", numbers[-1])
print("Total:", sum(numbers))
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))


# 9. LIST OPERATIONS

print("\n9. LIST OPERATIONS")

numbers.append(60)
print("After Append:", numbers)

numbers.remove(20)
print("After Remove:", numbers)


# 10. TUPLE

print("\n10. TUPLE")

fruits = ("Apple", "Mango", "Banana")

print("Fruits:", fruits)
print("First Fruit:", fruits[0])


# 11. DICTIONARY

print("\n11. DICTIONARY")

student = {
    "name": "Subhasis",
    "age": 22,
    "course": "AI & ML"
}

print("Student:", student)
print("Name:", student["name"])
print("Course:", student["course"])


# 12. FUNCTIONS

print("\n12. FUNCTIONS")

def add(a, b):
    return a + b

result = add(10, 20)

print("Addition Result:", result)


# 13. FUNCTION FOR AVERAGE

def calculate_average(marks):
    return sum(marks) / len(marks)

student_marks = [80, 70, 90, 85]

average = calculate_average(student_marks)

print("Average Marks:", average)


# 14. STRING OPERATIONS

print("\n14. STRING OPERATIONS")

text = "Artificial Intelligence"

print("Original:", text)
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Length:", len(text))


# 15. SIMPLE DATA ANALYSIS

print("\n15. SIMPLE DATA ANALYSIS")

scores = [75, 80, 90, 65, 85]

print("Scores:", scores)
print("Total Score:", sum(scores))
print("Average Score:", sum(scores) / len(scores))
print("Highest Score:", max(scores))
print("Lowest Score:", min(scores))


# 16. FINAL OUTPUT

print("\n===== PROJECT COMPLETED =====")
print("Module 01 Python Practice Completed!")
```
